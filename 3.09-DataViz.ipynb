#### Q5.

Vamos fazer um compilado com as estatísticas históricas de copas do mundo!

Com a biblioteca de sua preferência você deve criar 4 subplots organizados em um grid de 2 linhas e 2 colunas. Eles devem conter os seguintes gráficos:
* Linha 1, coluna 1: Gráfico de barras com a quantidade de jogos que aconteceram por ano
* Linha 1, coluna 2: Gráfico de área (referências a seguir) com o total de gols por ano, separando as informações de `gols_1` e `gols_2` para distinguir gols em casa e do time visitante.
* Linha 2, coluna 1: Gráfico de área com o total de cartões por ano, separando as informações de cartões amarelos e cartões vermelhos, mas agregando cartões do time 1 ou time 2. Ou seja, uma área com `cartao_amarelo_1 + cartao_amarelo_2` e outra área com `cartao_vermelho_1 + cartao_vermelho_2`.
* Linha 2, coluna 2: Gráfico de barras com o total de gols contra por ano, somando `gols_contra_1` e `gols_contra_2`.

Referências sobre gráfico de área
* Matplotlib: https://matplotlib.org/stable/gallery/lines_bars_and_markers/stackplot_demo.html#sphx-glr-gallery-lines-bars-and-markers-stackplot-demo-py
* Pandas + Matplotlib: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.area.html
* Seaborn: https://seaborn.pydata.org/generated/seaborn.objects.Area.html
* Plotly: https://plotly.com/python/filled-area-plots/

````python
import ast

# Função para contabilizar ocorrências em strings contendo listas/separadores
def contar_itens(val):
    if pd.isna(val) or not val:
        return 0
    if isinstance(val, str):
        try:
            l = ast.literal_eval(val)
            if isinstance(l, list):
                return len(l)
        except Exception:
            pass
        if '|' in val:
            return val.count('|') // 2
    return 0

# Contagem prévia de eventos
wc['num_ca1'] = wc['cartao_amarelo_1'].apply(contar_itens)
wc['num_ca2'] = wc['cartao_amarelo_2'].apply(contar_itens)
wc['num_cv1'] = wc['cartao_vermelho_1'].apply(contar_itens)
wc['num_cv2'] = wc['cartao_vermelho_2'].apply(contar_itens)

wc['num_gc1'] = wc['gols_1_contra'].apply(contar_itens)
wc['num_gc2'] = wc['gols_2_contra'].apply(contar_itens)

# Agrupamento anual de estatísticas
estat_ano = wc.groupby('ano').apply(lambda g: pd.Series({
    'total_jogos': len(g),
    'gols_mandante': g['gols_1'].sum(),
    'gols_visitante': g['gols_2'].sum(),
    'cartoes_amarelos': g['num_ca1'].sum() + g['num_ca2'].sum(),
    'cartoes_vermelhos': g['num_cv1'].sum() + g['num_cv2'].sum(),
    'gols_contra': g['num_gc1'].sum() + g['num_gc2'].sum()
})).reset_index()

# Criando Grid 2x2 com Matplotlib
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# L1, C1: Gráfico de barras com a quantidade de jogos por ano
axes[0, 0].bar(estat_ano['ano'], estat_ano['total_jogos'], color='navy')
axes[0, 0].set_title('Quantidade de Jogos por Ano')
axes[0, 0].set_xlabel('Ano')
axes[0, 0].set_ylabel('Nº de Jogos')

# L1, C2: Gráfico de área com o total de gols por ano (gols_1 vs gols_2)
axes[0, 1].stackplot(estat_ano['ano'], estat_ano['gols_mandante'], estat_ano['gols_visitante'],
                     labels=['Gols Casa (Mandante)', 'Gols Fora (Visitante)'], colors=['tab:blue', 'tab:orange'], alpha=0.7)
axes[0, 1].set_title('Total de Gols por Ano (Mandante vs Visitante)')
axes[0, 1].set_xlabel('Ano')
axes[0, 1].set_ylabel('Total de Gols')
axes[0, 1].legend(loc='upper left')

# L2, C1: Gráfico de área com cartões amarelos vs vermelhos
axes[1, 0].stackplot(estat_ano['ano'], estat_ano['cartoes_amarelos'], estat_ano['cartoes_vermelhos'],
                     labels=['Cartões Amarelos', 'Cartões Vermelhos'], colors=['gold', 'crimson'], alpha=0.7)
axes[1, 0].set_title('Total de Cartões por Ano')
axes[1, 0].set_xlabel('Ano')
axes[1, 0].set_ylabel('Quantidade de Cartões')
axes[1, 0].legend(loc='upper left')

# L2, C2: Gráfico de barras com total de gols contra por ano
axes[1, 1].bar(estat_ano['ano'], estat_ano['gols_contra'], color='darkred')
axes[1, 1].set_title('Total de Gols Contra por Ano')
axes[1, 1].set_xlabel('Ano')
axes[1, 1].set_ylabel('Gols Contra')

plt.tight_layout()
plt.show()
````
